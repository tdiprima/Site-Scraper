# BeautifulSoup Web Scraping Tutorial

````python
from urllib.request import urlopen
from bs4 import BeautifulSoup
````

````python
url = "http://example.com"

# Open and read the URL
response = urlopen(url)
html = response.read()

# Parse the HTML with BeautifulSoup
soup = BeautifulSoup(html, 'html.parser')
````

````python
print(soup.prettify())
````

````python
headers = soup.find_all('h1')
for header in headers:
    print(header.get_text())
````

````python
# Select elements by CSS selectors
links = soup.select('a')

for link in links:
    href = link.get('href')
    text = link.get_text()
    print(f"Link text: {text}, URL: {href}")
````

````python
# By class
elements_with_class = soup.find_all('div', class_='myClass')

# By id
element_with_id = soup.find(id='uniqueId')

# By attribute
images = soup.find_all('img', alt='image description')

elements_with_class, element_with_id, images
````

## Extracting news titles from Hacker News

```html
<span class="titleline"><a href="https://federicopereiro.com/fat-trap/">Being fat is a trap</a><span class="sitebit comhead"> (<a href="from?site=federicopereiro.com"><span class="sitestr">federicopereiro.com</span></a>)</span></span>
```

````python
# Extracting news titles from Hacker News

url = "https://news.ycombinator.com"
response = urlopen(url)
html = response.read()

soup = BeautifulSoup(html, 'html.parser')

titles = soup.select('.titleline')
# Use enumerate() when you want both the index and the item.
for idx, title in enumerate(titles, start=1):
    headline = title.find('a').get_text()
    print(f"{idx}: {headline}")
````

## Check robots.txt Before Scraping
